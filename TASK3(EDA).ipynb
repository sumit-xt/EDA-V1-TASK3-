```python
import pandas as pd 
data=pd.read_csv("hotel_booking.csv")
print(data.head(10))
missingdata=data.isnull().sum()
print("________________________missingdata____________________")
print(missingdata)
dupdata=data.duplicated().sum()
print("________________________duplicate data____________________")
print(dupdata)
print("____________________________________________")
rowscol=data.shape
print("________________________rows and columns____________________")
print(rowscol)
print("____________________________________________")
info1=data.info()
print(info1)
print("_____________________stats related info_______________________")
print(data.describe())
import matplotlib.pyplot as plt
data["hotel"].value_counts().plot(kind="bar")
plt.title(" no of hotel-booking count")
plt.xlabel("hotel type")
plt.ylabel("no of booking")
plt.show()
plt.hist(data["lead_time"],bins=20)
plt.title("distribution of how before booking are done")
plt.xlabel("lead time")
plt.ylabel("frequency")
plt.show()








```
